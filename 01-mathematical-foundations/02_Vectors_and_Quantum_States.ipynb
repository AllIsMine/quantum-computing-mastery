```python
# Cell 1: Environment setup

import sys
import numpy as np

print("Python version:", sys.version)
print("NumPy version:", np.__version__)
print("Environment setup successful.")
```

```python
# Cell 2: Basic vector representation

v = np.array([3, 4], dtype=float)

print("Vector:", v)
print("First component:", v[0])
print("Second component:", v[1])

assert np.allclose(v, [3, 4])
print("Vector representation check passed.")
```

```python
# Cell 3: Computational basis states

zero = np.array([1, 0], dtype=complex)
one = np.array([0, 1], dtype=complex)

print("|0> =", zero)
print("|1> =", one)

assert np.allclose(zero, [1, 0])
assert np.allclose(one, [0, 1])

print("Computational basis-state checks passed.")
```

```python
# Cell 4: Norms and orthogonality

norm_zero = np.linalg.norm(zero)
norm_one = np.linalg.norm(one)
inner_product = np.vdot(zero, one)

print("Norm of |0>:", norm_zero)
print("Norm of |1>:", norm_one)
print("<0|1>:", inner_product)

assert np.isclose(norm_zero, 1.0)
assert np.isclose(norm_one, 1.0)
assert np.isclose(inner_product, 0.0)

print("Norm and orthogonality checks passed.")
```

```python
# Cell 5: Measurement probabilities

psi = np.array([3/5, 4/5], dtype=complex)
probabilities = np.abs(psi) ** 2

print("State:", psi)
print("P(0):", probabilities[0])
print("P(1):", probabilities[1])
print("Total probability:", probabilities.sum())

assert np.allclose(probabilities, [0.36, 0.64])
assert np.isclose(probabilities.sum(), 1.0)

print("Measurement-probability checks passed.")
```

```python
# Cell 6: Norm and squared norm

v = np.array([3.0, 4.0])

norm_v = np.linalg.norm(v)
squared_norm = norm_v ** 2

print("Vector:", v)
print("Norm:", norm_v)
print("Squared norm:", squared_norm)

assert np.isclose(norm_v, 5.0)
assert np.isclose(squared_norm, 25.0)

print("Norm checks passed.")
```

```python
# Cell 7: Real-vector normalization

v = np.array([3.0, 4.0])
norm_v = np.linalg.norm(v)

if np.isclose(norm_v, 0.0):
    raise ValueError("The zero vector cannot be normalized.")

psi = v / norm_v

print("Original vector:", v)
print("Original norm:", norm_v)
print("Normalized vector:", psi)
print("Normalized norm:", np.linalg.norm(psi))

assert np.allclose(psi, [0.6, 0.8])
assert np.isclose(np.linalg.norm(psi), 1.0)

print("Real-vector normalization checks passed.")
```

```python
# Cell 8: Complex-vector normalization

v = np.array([1 + 1j, 1 - 1j], dtype=complex)
norm_v = np.linalg.norm(v)

if np.isclose(norm_v, 0.0):
    raise ValueError("The zero vector cannot be normalized.")

psi = v / norm_v
probabilities = np.abs(psi) ** 2

print("Original vector:", v)
print("Original norm:", norm_v)
print("Normalized state:", psi)
print("Component probabilities:", probabilities)
print("Total probability:", probabilities.sum())

assert np.isclose(norm_v, 2.0)
assert np.allclose(
    psi,
    [(1 + 1j) / 2, (1 - 1j) / 2]
)
assert np.allclose(probabilities, [0.5, 0.5])
assert np.isclose(np.linalg.norm(psi), 1.0)

print("Complex-state normalization checks passed.")
```

```python
# Cell 9: Equal superposition

plus = np.array(
    [1 / np.sqrt(2), 1 / np.sqrt(2)],
    dtype=complex
)

probabilities = np.abs(plus) ** 2

print("|+> =", plus)
print("Norm:", np.linalg.norm(plus))
print("Probabilities:", probabilities)
print("Total probability:", probabilities.sum())

assert np.isclose(np.linalg.norm(plus), 1.0)
assert np.allclose(probabilities, [0.5, 0.5])
assert np.isclose(probabilities.sum(), 1.0)

print("Equal-superposition checks passed.")
```

```python
# Cell 10: Unequal measurement probabilities

psi = np.array(
    [1 / np.sqrt(5), 2 / np.sqrt(5)],
    dtype=complex
)

probabilities = np.abs(psi) ** 2

print("State:", psi)
print("Probabilities:", probabilities)
print("Total probability:", probabilities.sum())

assert np.allclose(probabilities, [0.2, 0.8])
assert np.isclose(np.linalg.norm(psi), 1.0)
assert np.isclose(probabilities.sum(), 1.0)

print("Unequal-probability state checks passed.")
```

```python
# Cell 11: Challenge A

v = np.array([5.0, 12.0])
norm_v = np.linalg.norm(v)

if np.isclose(norm_v, 0.0):
    raise ValueError("The zero vector cannot be normalized.")

psi = v / norm_v

print("Original vector:", v)
print("Norm:", norm_v)
print("Normalized vector:", psi)
print("Normalized norm:", np.linalg.norm(psi))

assert np.isclose(norm_v, 13.0)
assert np.allclose(psi, [5/13, 12/13])
assert np.isclose(np.linalg.norm(psi), 1.0)

print("Challenge A passed.")
```

```python
# Cell 12: Challenge B

v = np.array([1 + 1j, 1 - 1j], dtype=complex)
norm_v = np.linalg.norm(v)

if np.isclose(norm_v, 0.0):
    raise ValueError("The zero vector cannot be normalized.")

psi = v / norm_v
probabilities = np.abs(psi) ** 2

print("Original norm:", norm_v)
print("Normalized state:", psi)
print("Probabilities:", probabilities)
print("Total probability:", probabilities.sum())

assert np.isclose(norm_v, 2.0)
assert np.allclose(probabilities, [0.5, 0.5])
assert np.isclose(np.linalg.norm(psi), 1.0)
assert np.isclose(probabilities.sum(), 1.0)

print("Challenge B passed.")
```

```python
# Cell 13: Challenge C

psi = np.array(
    [np.sqrt(3) / 2, 1 / 2],
    dtype=complex
)

probabilities = np.abs(psi) ** 2

print("State:", psi)
print("P(0):", probabilities[0])
print("P(1):", probabilities[1])
print("Total probability:", probabilities.sum())

assert np.allclose(probabilities, [0.75, 0.25])
assert np.isclose(probabilities.sum(), 1.0)
assert np.isclose(np.linalg.norm(psi), 1.0)

print("Challenge C passed.")
```

```python
# Cell 14: Challenge D

v = np.array([1.0, 1.0])
norm_v = np.linalg.norm(v)

if np.isclose(norm_v, 0.0):
    raise ValueError("The zero vector cannot be normalized.")

psi = v / norm_v

print("Original norm:", norm_v)
print("Normalized vector:", psi)
print("Normalized norm:", np.linalg.norm(psi))

assert np.isclose(norm_v, np.sqrt(2))
assert np.allclose(psi, [1/np.sqrt(2), 1/np.sqrt(2)])
assert np.isclose(np.linalg.norm(psi), 1.0)

print("Challenge D passed.")
```

```python
# Cell 15: Challenge E

psi1 = np.array(
    [1 / np.sqrt(2), 1 / np.sqrt(2)],
    dtype=complex
)

psi2 = np.array(
    [1 / np.sqrt(2), -1 / np.sqrt(2)],
    dtype=complex
)

probabilities1 = np.abs(psi1) ** 2
probabilities2 = np.abs(psi2) ** 2

print("State 1:", psi1)
print("State 2:", psi2)
print("Probabilities of state 1:", probabilities1)
print("Probabilities of state 2:", probabilities2)
print("Inner product <psi1|psi2>:", np.vdot(psi1, psi2))

assert np.allclose(probabilities1, [0.5, 0.5])
assert np.allclose(probabilities2, [0.5, 0.5])
assert not np.allclose(psi1, psi2)
assert np.isclose(np.vdot(psi1, psi2), 0.0)

print("Challenge E numerical checks passed.")
```
